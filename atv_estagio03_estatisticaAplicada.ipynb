Base de dados:
- Número de observações(linhas): 1997

Variáveis:
- tipo_servico: Corresponde ao tipo de serviço contratado/prestado ao cliente, é uma unidade do tipo qualitativo categórica 
- canal_atendimento: Corresponde ao meio de contato utilizado pelo cliente, é uma unidade do tipo qualitativo categórica
- faixa_etaria: Corresponde a faixa de idade do cliente, é uma unidade do tipo qualitativa ordinal
- tempo_espera: Corresponde ao tempo esperado pelo cliente para ser atendido, é uma unidade do tipo quantitativa contínua (em minutos)
- qualidade_percebida: Corresponde a um valor de avaliação do cliente pelo atendimento recebido, é uma unidade quantitativa contínua de 0 a 10
- preco_pago: Corresponde ao valor pago pelo cliente pelo serviço prestado, é uma unidade do tipo quantitativa contínua (em reais)
- atendimento_resolutivo: Corresponde a uma resposta se um atendimento resolveu ou não a questão do cliente, é uma unidade qualitativa binária (pode ser tratado como boolean)
- nps: Corresponde am um indice de lealdade do cliente, ou seja, uma probabilidade do cliente recomendar esse serviço, é uma unidade quantitativa contínua de 0 a 10
- tempo_resposta: Corresponde ao tempo que a empresa levou para atender o cliente, é uma unidade do tipo quantitativa contínua (em minutos)
- isc: Corresponde a um indicador de satisfação geral do cliente, é uma unidade do tipo quantitativa contínua (em percentual)

#### Organizando dados

```python
df_SatisfacaoCliente
```

```python
# Passando numeros do formato brasileiro para o americano ("0,0" -> "0.0")

df_SatisfacaoCliente["tempo_espera"] = (df_SatisfacaoCliente["tempo_espera"].str.replace(",", ".", regex=False).astype(float))

df_SatisfacaoCliente["qualidade_percebida"] = (df_SatisfacaoCliente["qualidade_percebida"].str.replace(",", ".", regex=False).astype(float))

df_SatisfacaoCliente["preco_pago"] = (df_SatisfacaoCliente["preco_pago"].str.replace(",", ".", regex=False).astype(float))

df_SatisfacaoCliente["nps"] = (df_SatisfacaoCliente["nps"].str.replace(",", ".", regex=False).astype(float))

df_SatisfacaoCliente["tempo_resposta"] = (df_SatisfacaoCliente["tempo_resposta"].str.replace(",", ".", regex=False).astype(float))

df_SatisfacaoCliente["isc"] = (df_SatisfacaoCliente["isc"].str.replace(",", ".", regex=False).astype(float))
```

```python
df_SatisfacaoCliente
```

```python
#Passando variaveis quantitativas de string para float

df_SatisfacaoCliente.tempo_espera = pd.to_numeric(df_SatisfacaoCliente.tempo_espera, errors = 'coerce')
df_SatisfacaoCliente.qualidade_percebida = pd.to_numeric(df_SatisfacaoCliente.qualidade_percebida, errors = 'coerce')
df_SatisfacaoCliente.preco_pago= pd.to_numeric(df_SatisfacaoCliente.preco_pago, errors = 'coerce')
df_SatisfacaoCliente.nps = pd.to_numeric(df_SatisfacaoCliente.nps, errors = 'coerce')
df_SatisfacaoCliente.tempo_resposta = pd.to_numeric(df_SatisfacaoCliente.tempo_resposta, errors = 'coerce')
df_SatisfacaoCliente.isc = pd.to_numeric(df_SatisfacaoCliente.isc, errors = 'coerce')
```
